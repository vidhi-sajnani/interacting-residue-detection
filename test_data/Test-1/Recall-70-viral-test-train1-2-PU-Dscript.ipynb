```python
import pandas as pd
from sklearn.metrics import recall_score
```

```python
# Load the CSV file
file_path = 'viral-test-recall-M1.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of M1 with Train 1:", recall1)
print("Recall of M1 with Train 2:", recall2)


# Load the CSV file
file_path = 'viral-test-recall-NEP.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of NEP with Train 1:", recall1)
print("Recall of NEP with Train 2:", recall2)


# Load the CSV file
file_path = 'viral-test-recall-NP.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of NP with Train 1:", recall1)
print("Recall of NP with Train 2:", recall2)


# Load the CSV file
file_path = 'viral-test-recall-NS1.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of NS1 with Train 1:", recall1)
print("Recall of NS1 with Train 2:", recall2)

# Load the CSV file
file_path = 'viral-test-recall-PB1.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of PB1 with Train 1:", recall1)
print("Recall of PB1 with Train 2:", recall2)


# Load the CSV file
file_path = 'viral-test-recall-PB2.csv'
df = pd.read_csv(file_path)

# Replace non-1 and NaN values with 0 in Label
df['Label'] = df['Label'].apply(lambda x: 1 if x == 1 else 0)

# Calculate recall
recall1 = recall_score(df['Label'], df['Train1-RandomSampler'])
recall2 = recall_score(df['Label'], df['Train2-RandomSampler'])
print("Recall of PB2 with Train 1:", recall1)
print("Recall of PB2 with Train 2:", recall2)
```
